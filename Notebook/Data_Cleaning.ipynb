# 2. Data Cleaning


## Overview

Tahap data cleaning dilakukan untuk memastikan dataset memiliki kualitas yang baik, konsisten, dan siap digunakan untuk proses analisis selanjutnya.

Pada tahap ini dilakukan identifikasi, investigasi, dan penanganan terhadap berbagai permasalahan kualitas data, seperti:

* Duplicate data
* Missing value
* Kesalahan data type
* Invalid value
* Outlier

Setiap permasalahan pada data tidak langsung dihapus atau diubah. Terlebih dahulu dilakukan investigasi untuk menentukan apakah data tersebut merupakan data yang valid atau merupakan kesalahan data.

### Data Cleaning Workflow

**Duplicate Detection**
↓
**Duplicate Investigation**
↓
**Duplicate Treatment**
↓
**Missing Value Detection**
↓
**Missing Value Investigation**
↓
**Missing Value Treatment**
↓
**Data Type Validation**
↓
**Data Type Conversion**
↓
**Invalid Value Detection**
↓
**Invalid Value Investigation**
↓
**Invalid Value Treatment**
↓
**Outlier Detection**
↓
**Outlier Investigation**
↓
**Outlier Treatment**
↓
**Data Validation**
↓
**Clean Dataset**

### Objective

Tujuan utama tahap data cleaning adalah:

1. Mengidentifikasi permasalahan pada dataset.
2. Menentukan apakah data yang bermasalah valid atau tidak.
3. Menangani data yang tidak sesuai dengan kebutuhan analisis.
4. Memastikan konsistensi tipe data dan nilai pada setiap kolom.
5. Memastikan dataset siap digunakan untuk tahap feature engineering, EDA, dan business analysis.


```python
import pandas as pd 
import matplotlib.pyplot as plt 

pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv("../Dataset/ecommerce_retail_dataset.csv")
df.head(10)
```

1. Menghapus Data duplikat 

```python
df[df.duplicated(keep=False)]
```

```python
data_clean = df.drop_duplicates()
data_clean.duplicated().sum()
```

2. Mising Value

```python
data_clean.isnull().sum()
```

```python
data_clean[data_clean["delivery_date"].isnull()]
```

```python
df.loc[
    df["delivery_date"].isna(),
    ["order_id", "order_date", "customer_id","order_status", "delivery_date"]
]
```

```python
df[df["rating"].isna()]
```

```python
data_clean["customer_city"] = data_clean["customer_city"].fillna(data_clean["customer_city"].mode()[0])
data_clean["rating"] = data_clean["rating"].fillna(data_clean["rating"].median())
```

3. Data Type

```python
data_clean["order_date"] = pd.to_datetime(df["order_date"])
data_clean["delivery_date"] = pd.to_datetime(df["delivery_date"])
```

4. Outlier 

```python
Q1 = data_clean["unit_price"].quantile(0.25)
Q3 = data_clean["unit_price"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_price = data_clean[(data_clean["unit_price"] <  lower_bound) | (data_clean["unit_price"] > upper_bound)]
outlier_price
```

```python
outlier_price[
    [
        "order_id",
        "product_name",
        "category",
        "quantity",
        "unit_price",
        "total_amount"
    ]
]
```

```python
Q1 = data_clean["quantity"].quantile(0.25)
Q3 = data_clean["quantity"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_quantity = data_clean[(data_clean["quantity"] <  lower_bound) | (data_clean["quantity"] > upper_bound)]
outlier_quantity
```

```python
data_clean["quantity"].describe()
```

```python
outlier_quantity[
    [
        "order_id",
        "product_name",
        "category",
        "quantity",
        "unit_price",
        "total_amount"
    ]
]
```

```python
Q1 = data_clean["total_amount"].quantile(0.25)
Q3 = data_clean["total_amount"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_amount = data_clean[(data_clean["total_amount"] <  lower_bound) | (data_clean["total_amount"] > upper_bound)]
outlier_amount
```

```python
outlier_amount[
    [
        "order_id",
        "product_name",
        "category",
        "unit_price",
        "quantity",
        "discount_percent",
        "shipping_cost",
        "total_amount"
    ]
]
```
