# 3. Feature Engineering


Feature engineering dilakukan untuk membuat variabel baru yang dapat membantu proses analisis dan memberikan informasi tambahan dari data yang sudah melalui tahap data cleaning.


```python
import pandas as pd 

pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv("../Dataset/data_cleaning.csv")
df.head()
```

1. Time Features

```python
df["order_date"] = pd.to_datetime(df["order_date"])
df["delivery_date"] = pd.to_datetime(df["delivery_date"])
```

```python
df["order_month"] = df["order_date"].dt.month
df["order_year"] = df["order_date"].dt.year
df["order_month_name"] = df["order_date"].dt.month_name()
```

2. Transaction Features

```python
df["gross_amount"] = df["unit_price"] * df["quantity"]
df["discount_amount"] = (df["gross_amount"] * df["discount_percent"] / 100)
```

3. Delivery Features

```python
df["delivery_days"] = (df["delivery_date"] - df["order_date"]).dt.days
```

4. Customer Feature

```python
df["customer_age_group"] = pd.cut(
    df["customer_age"],
    bins=[0, 19, 29, 39, 49, 59, float("inf")],
    labels=["<20", "20-29", "30-39", "40-49", "50-59", "60+"]
)
```
